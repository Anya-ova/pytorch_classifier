```python
# ФИНАЛЬНЫЙ МАППИНГ на основе реальных предсказаний модели
IMAGENET_CLASS_MAPPING = {
    'drilling_rig': [
        # Основные правильные предсказания
        'drilling platform',  # ✅ ЛУЧШЕЕ СОВПАДЕНИЕ!
        # Структурные аналоги
        'water tower',
        'oil refinery',
        'tower'
    ],
    'pumping_unit': [
        'crane',
        # Механические аналоги
        'gas pump',
        'water pump'
    ]
}
```

```python
def is_correct_prediction(true_class, predicted_class, confidence_threshold=0.0):
    """
    Проверяет, правильно ли модель распознала класс
    на основе маппинга классов ImageNet
    """
    if true_class not in IMAGENET_CLASS_MAPPING:
        return False

    predicted_class_lower = predicted_class.lower().strip()

    # Проверяем точное совпадение
    for valid_class in IMAGENET_CLASS_MAPPING[true_class]:
        if valid_class.lower() == predicted_class_lower:
            return True

    # Проверяем частичное совпадение
    for valid_class in IMAGENET_CLASS_MAPPING[true_class]:
        if valid_class.lower() in predicted_class_lower:
            return True
        if predicted_class_lower in valid_class.lower():
            return True

    return False
```

```python
def get_device(device_type='auto'):
    """Определяет доступное устройство"""
    if device_type == 'tpu' and HAS_TPU:
        device = xm.xla_device()
        print(f"Используется TPU: {device}")
        return device
    elif device_type == 'cuda' and torch.cuda.is_available():
        device = torch.device("cuda:0")
        print(f"Используется GPU: {device}")
        return device
    elif device_type == 'auto':
        if HAS_TPU:
            device = xm.xla_device()
            print(f"Автовыбор: TPU: {device}")
            return device
        elif torch.cuda.is_available():
            device = torch.device("cuda:0")
            print(f"Автовыбор: GPU: {device}")
            return device
        else:
            device = torch.device("cpu")
            print(f"Автовыбор: CPU: {device}")
            return device
    else:
        device = torch.device("cpu")
        print(f"Используется CPU: {device}")
        return device
```

```python
class AugmentedFolderDataset(Dataset):
    """Dataset с поддержкой аугментации"""
    def __init__(self, data_paths, transform=None, augment=True):
        self.transform = transform
        self.augment = augment
        self.samples = []
        self.classes = ['drilling_rig', 'pumping_unit']  # Исправлено: pumping_unit
        self.class_to_idx = {cls: idx for idx, cls in enumerate(self.classes)}
        self.idx_to_class = {idx: cls for cls, idx in self.class_to_idx.items()}

        for class_name in self.classes:
            class_path = data_paths[class_name]
            if os.path.exists(class_path):
                for img_name in os.listdir(class_path):
                    if img_name.lower().endswith(('.png', '.jpg', '.jpeg', '.bmp')):
                        img_path = os.path.join(class_path, img_name)
                        self.samples.append((img_path, self.class_to_idx[class_name]))

        print(f"Загружено {len(self.samples)} изображений")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        try:
            img = Image.open(img_path).convert('RGB')

            # Применяем аугментацию если нужно
            if self.augment and self.transform:
                img = self.transform(img)
            elif self.transform:
                img = self.transform(img)

            return img, label, img_path
        except Exception as e:
            print(f"Ошибка загрузки {img_path}: {e}")
            # Возвращаем пустое изображение в случае ошибки
            dummy_img = Image.new('RGB', (224, 224), color='black')
            if self.transform:
                dummy_img = self.transform(dummy_img)
            return dummy_img, label, img_path
```

```python
def test_actual_predictions(num_samples=10):
    """Тестирует, что реально предсказывает модель"""
    print("=== ТЕСТ РЕАЛЬНЫХ ПРЕДСКАЗАНИЙ МОДЕЛИ ===")

    model_info = models_dict['AlexNet']
    model = model_info['model'].to(device)
    model.eval()
    imagenet_classes = model_info['weights'].meta["categories"]

    results = {'drilling_rig': {}, 'pumping_unit': {}}

    with torch.no_grad():
        for idx in range(min(num_samples, len(dataset))):
            img, label, path = dataset[idx]
            img_tensor = img.unsqueeze(0).to(device)

            output = model(img_tensor)
            probs = torch.nn.functional.softmax(output, dim=1)
            top5_conf, top5_idx = torch.topk(probs, 5)

            true_class = dataset.idx_to_class[label]

            print(f"\n{idx+1}. {true_class}: {path.split('/')[-1]}")
            print("Топ-5 предсказаний:")

            for j in range(5):
                predicted = imagenet_classes[top5_idx[0][j].item()]
                confidence = top5_conf[0][j].item()
                print(f"  {j+1}. {predicted} ({confidence:.1%})")

                # Сохраняем статистику
                if predicted not in results[true_class]:
                    results[true_class][predicted] = 0
                results[true_class][predicted] += 1

    # Выводим статистику
    print("\n=== СТАТИСТИКА ПРЕДСКАЗАНИЙ ===")
    for class_name, pred_stats in results.items():
        print(f"\n{class_name}:")
        sorted_stats = sorted(pred_stats.items(), key=lambda x: x[1], reverse=True)
        for pred, count in sorted_stats[:10]:  # топ-10 предсказаний
            print(f"  {pred}: {count} раз")

# Запустите тест
test_actual_predictions(50)
```

```python
def test_model_performance(model_info, dataloader, device, device_type='cpu', num_samples=None):
    """
    Тестирование скорости, точности и реальной успешности модели
    """
    model = copy.deepcopy(model_info["model"])
    model = model.to(device)
    model.eval()

    inference_times = []
    all_preds_random = []
    all_labels = []
    correct_predictions_real = 0
    total_predictions = 0
    imagenet_classes = model_info['weights'].meta["categories"]

    # Разогрев модели
    with torch.no_grad():
        dummy_input = torch.randn(1, 3, 224, 224).to(device)
        if device_type == 'cuda':
            torch.cuda.synchronize()
        _ = model(dummy_input)
        if device_type == 'cuda':
            torch.cuda.synchronize()

    with torch.no_grad():
        batch_count = 0
        for images, labels, _ in dataloader:
            if num_samples and batch_count >= num_samples:
                break

            images = images.to(device)
            labels = labels.to(device)

            # Измеряем время инференса
            if device_type == 'cuda':
                torch.cuda.synchronize()
                start_time = time.perf_counter()

                outputs = model(images)
                torch.cuda.synchronize()
                end_time = time.perf_counter()
            else:
                start_time = time.perf_counter()
                outputs = model(images)
                end_time = time.perf_counter()

            inference_times.append((end_time - start_time) * 1000)  # В мс

            # Получаем вероятности
            probs = torch.nn.functional.softmax(outputs, dim=1)

            # Получаем топ-5 предсказаний
            top5_conf, top5_idx = torch.topk(probs, 5)

            for i in range(len(labels)):
                true_class = dataset.idx_to_class[labels[i].item()]
                found_correct = False

                # Проверяем топ-5 предсказаний на соответствие маппингу
                for j in range(5):
                    predicted_class = imagenet_classes[top5_idx[i][j].item()]
                    confidence = top5_conf[i][j].item()

                    if is_correct_prediction(true_class, predicted_class):
                        correct_predictions_real += 1
                        found_correct = True
                        break

                total_predictions += 1

                # Для случайной точности берем самое вероятное предсказание
                pred_idx = torch.argmax(outputs[i]).item()
                all_preds_random.append(pred_idx)
                all_labels.append(labels[i].item())

            batch_count += 1

    # Вычисляем метрики
    avg_time = np.mean(inference_times) if inference_times else 0
    random_accuracy = accuracy_score(all_labels, all_preds_random) if all_preds_random else 0
    real_accuracy = correct_predictions_real / total_predictions if total_predictions > 0 else 0

    return avg_time, random_accuracy, real_accuracy
```

```python
test_loader = DataLoader(dataset, batch_size=1, shuffle=False, num_workers=0)

# Словарь для хранения результатов
results = {
    'cpu': {'times': {}, 'random_acc': {}, 'real_acc': {}},
    'gpu': {'times': {}, 'random_acc': {}, 'real_acc': {}},
    'tpu': {'times': {}, 'random_acc': {}, 'real_acc': {}}
}

print("\n Тестирование на CPU:")
for model_name, model_info in models_dict.items():
    time_ms, random_acc, real_acc = test_model_performance(
        model_info, test_loader, device, 'cpu')
    results['cpu']['times'][model_name] = time_ms
    results['cpu']['random_acc'][model_name] = random_acc
    results['cpu']['real_acc'][model_name] = real_acc
    print(f"  {model_name:15} : {time_ms:6.1f} мс | "
          f"Случайная: {random_acc:6.2%} | Реальная: {real_acc:6.2%}")

# ТЕСТИРОВАНИЕ НА GPU (ЕСЛИ ДОСТУПНО)
if torch.cuda.is_available():

    gpu_device = torch.device("cuda:0")
    print(f"\n Тестирование на GPU:")

    for model_name, model_info in models_dict.items():
        time_ms, random_acc, real_acc = test_model_performance(
            model_info, test_loader, gpu_device, 'cuda')
        results['gpu']['times'][model_name] = time_ms
        results['gpu']['random_acc'][model_name] = random_acc
        results['gpu']['real_acc'][model_name] = real_acc
        print(f"  {model_name:15} : {time_ms:6.1f} мс | "
              f"Случайная: {random_acc:6.2%} | Реальная: {real_acc:6.2%}")
```

```python
all_examples = {}

for model_name, model_info in models_dict.items():
    model = model_info['model'].to(device)
    model.eval()
    imagenet_classes = model_info['weights'].meta["categories"]

    examples = {
        'success': {0: None, 1: None},  # Успешные примеры для каждого класса
        'failure': {0: None, 1: None}   # Неуспешные примеры для каждого класса
    }

    success_found = {0: False, 1: False}
    failure_found = {0: False, 1: False}

    # Собираем все результаты для этой модели
    all_results = []

    with torch.no_grad():
        for idx in range(len(dataset)):
            img, label, path = dataset[idx]
            img_tensor = img.unsqueeze(0).to(device)

            # Получаем предсказания
            output = model(img_tensor)
            probs = torch.nn.functional.softmax(output, dim=1)
            top5_conf, top5_idx = torch.topk(probs, 5)

            true_class = dataset.idx_to_class[label]
            predicted_class = imagenet_classes[top5_idx[0][0].item()]
            confidence = top5_conf[0][0].item()

            # Проверяем, есть ли правильное предсказание в топ-5
            is_correct = False
            for j in range(5):
                pred_class = imagenet_classes[top5_idx[0][j].item()]
                if is_correct_prediction(true_class, pred_class):
                    is_correct = True
                    predicted_class = pred_class
                    confidence = top5_conf[0][j].item()
                    break

            all_results.append({
                'image': img,
                'true_label': label,
                'true_class': true_class,
                'pred_class': predicted_class,
                'confidence': confidence,
                'is_correct': is_correct,
                'path': path,
                'top5_predictions': [
                    (imagenet_classes[top5_idx[0][k].item()], top5_conf[0][k].item())
                    for k in range(min(5, len(imagenet_classes)))
                ]
            })

    # Теперь ищем лучшие примеры из собранных результатов
    # Для каждого класса ищем успешный и неуспешный примеры

    for label in [0, 1]:
        # Фильтруем результаты для текущего класса
        class_results = [r for r in all_results if r['true_label'] == label]

        if not class_results:
            continue

        # Ищем успешный пример (с наивысшей уверенностью)
        successful_results = [r for r in class_results if r['is_correct']]
        if successful_results:
            # Берем самый уверенный успешный пример
            best_success = max(successful_results, key=lambda x: x['confidence'])
            examples['success'][label] = best_success
            success_found[label] = True

        # Ищем неуспешный пример (с наивысшей уверенностью в неправильном предсказании)
        failed_results = [r for r in class_results if not r['is_correct']]
        if failed_results:
            # Берем самый уверенный неуспешный пример
            best_failure = max(failed_results, key=lambda x: x['confidence'])
            examples['failure'][label] = best_failure
            failure_found[label] = True

    # Если не нашли все примеры, используем любые доступные
    for label in [0, 1]:
        if not success_found[label] and class_results:
            # Используем первый доступный пример как "успешный" (даже если он неуспешный)
            examples['success'][label] = class_results[0]
            # Изменяем статус на успешный для отображения
            examples['success'][label]['is_correct'] = True

        if not failure_found[label] and class_results:
            # Используем последний доступный пример как "неуспешный"
            if len(class_results) > 1:
                examples['failure'][label] = class_results[-1]
            else:
                # Если только один пример, дублируем его как неуспешный
                examples['failure'][label] = copy.deepcopy(class_results[0])
                examples['failure'][label]['is_correct'] = False
                examples['failure'][label]['confidence'] = 0.3  # Низкая уверенность для неуспеха

    all_examples[model_name] = examples
    print(f"  {model_name}: успешных - {sum([1 for v in examples['success'].values() if v is not None])}, "
          f"неуспешных - {sum([1 for v in examples['failure'].values() if v is not None])}")
```

```python
def extract_features(model_info, dataloader, device):
    """Извлечение признаков из модели"""
    model = copy.deepcopy(model_info['model'])

    # Удаляем классификационный слой
    if hasattr(model, 'classifier'):
        if isinstance(model.classifier, nn.Sequential):
            model.classifier = nn.Sequential(*list(model.classifier.children())[:-1])
        else:
            model.classifier = nn.Identity()
    elif hasattr(model, 'fc'):
        model.fc = nn.Identity()
    elif hasattr(model, 'head'):
        model.head = nn.Identity()

    model = model.to(device)
    model.eval()

    all_features = []
    all_labels = []

    with torch.no_grad():
        for images, labels, _ in dataloader:
            images = images.to(device)
            features = model(images)

            # Если признаки многомерные, применяем global average pooling
            if len(features.shape) > 2:
                features = torch.mean(features, dim=[2, 3])

            all_features.append(features.cpu().numpy())
            all_labels.append(labels.numpy())

    if all_features:
        return np.vstack(all_features), np.concatenate(all_labels)
    return np.array([]), np.array([])
```

```python
def finetune_swint_model(dataset, train_ratio=0.8, epochs=15, learning_rate=0.0001, device='cpu'):
    """Донастройка модели SwinT на целевом датасете"""

    # Разделение датасета
    train_size = int(train_ratio * len(dataset))
    val_size = len(dataset) - train_size
    train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

    # Создание DataLoader'ов с аугментацией для тренировки
    train_dataset.dataset.augment = True
    train_dataset.dataset.transform = train_transform
    val_dataset.dataset.augment = False
    val_dataset.dataset.transform = val_transform

    train_loader = DataLoader(train_dataset, batch_size=4, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_dataset, batch_size=4, shuffle=False, num_workers=0)
    test_loader = DataLoader(dataset, batch_size=1, shuffle=False, num_workers=0)

    print(f"Обучающая выборка: {len(train_dataset)} изображений")
    print(f"Валидационная выборка: {len(val_dataset)} изображений")
    print(f"Всего эпох обучения: {epochs}")

    # Сохраняем оригинальную модель для сравнения признаков
    print("\n1. Подготовка модели для донастройки...")

    # Создаем две идентичные модели
    original_model = models.swin_t(weights=models.Swin_T_Weights.DEFAULT)
    model = models.swin_t(weights=models.Swin_T_Weights.DEFAULT)

    # Извлечение признаков ДО донастройки
    print("\n2. Извлечение признаков ДО донастройки...")
    original_features = []
    original_labels = []

    original_model.eval()
    original_model = original_model.to(device)

    # Создаем модель для извлечения признаков (без классификатора)
    feature_extractor_original = copy.deepcopy(original_model)
    # Сохраняем оригинальный head
    original_head = feature_extractor_original.head
    # Заменяем head на Identity для извлечения признаков
    feature_extractor_original.head = nn.Identity()
    feature_extractor_original = feature_extractor_original.to(device)
    feature_extractor_original.eval()

    with torch.no_grad():
        for images, labels, _ in test_loader:
            images = images.to(device)

            # Извлекаем признаки с помощью feature extractor
            features = feature_extractor_original(images)

            original_features.append(features.cpu().numpy())
            original_labels.append(labels.numpy())

    original_features = np.vstack(original_features) if original_features else np.array([])
    original_labels = np.concatenate(original_labels) if original_labels else np.array([])

    # Модификация модели для донастройки
    num_features = model.head.in_features

    # Сохраняем оригинальный head для извлечения признаков позже
    original_head = copy.deepcopy(model.head)

    # Заменяем финальный слой
    model.head = nn.Linear(num_features, 2)  # 2 выходных класса
    model = model.to(device)

    # Замораживаем все слои кроме финального
    for param in model.parameters():
        param.requires_grad = False
    for param in model.head.parameters():
        param.requires_grad = True

    # Настройка оптимизатора и функции потерь
    criterion = nn.CrossEntropyLoss()
    optimizer = optim.Adam(model.head.parameters(), lr=learning_rate)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=5, gamma=0.5)

    # Процесс обучения
    print("\n3. Обучение модели...")
    train_losses = []
    val_accuracies = []
    val_f1_scores = []

    for epoch in range(epochs):
        # Обучение
        model.train()
        epoch_train_loss = 0.0

        for images, labels, _ in train_loader:
            images = images.to(device)
            labels = labels.to(device)

            optimizer.zero_grad()
            outputs = model(images)
            loss = criterion(outputs, labels)
            loss.backward()
            optimizer.step()

            epoch_train_loss += loss.item()

        avg_train_loss = epoch_train_loss / len(train_loader)
        train_losses.append(avg_train_loss)

        # Валидация
        model.eval()
        correct = 0
        total = 0
        all_preds = []
        all_true = []

        with torch.no_grad():
            for images, labels, _ in val_loader:
                images = images.to(device)
                labels = labels.to(device)
                outputs = model(images)
                _, predicted = torch.max(outputs.data, 1)

                total += labels.size(0)
                correct += (predicted == labels).sum().item()

                all_preds.extend(predicted.cpu().numpy())
                all_true.extend(labels.cpu().numpy())

        val_accuracy = 100 * correct / total
        val_f1 = f1_score(all_true, all_preds, average='weighted') * 100
        val_accuracies.append(val_accuracy)
        val_f1_scores.append(val_f1)

        # Обновление learning rate
        scheduler.step()

        # Вывод прогресса
        if (epoch + 1) % 5 == 0 or epoch == 0 or epoch == epochs - 1:
            print(f"  Эпоха {epoch+1}/{epochs}: Потери = {avg_train_loss:.4f}, "
                  f"Точность = {val_accuracy:.2f}%, F1 = {val_f1:.2f}%")

    # Извлечение признаков ПОСЛЕ донастройки
    print("\n4. Извлечение признаков ПОСЛЕ донастройки...")
    model.eval()
    finetuned_features = []
    finetuned_labels = []

    # Создаем feature extractor из обученной модели
    feature_extractor_finetuned = copy.deepcopy(model)
    # Сохраняем обученный head
    saved_head = feature_extractor_finetuned.head
    # Заменяем head на Identity для извлечения признаков
    feature_extractor_finetuned.head = nn.Identity()
    feature_extractor_finetuned = feature_extractor_finetuned.to(device)
    feature_extractor_finetuned.eval()

    with torch.no_grad():
        for images, labels, _ in test_loader:
            images = images.to(device)

            # Извлекаем признаки
            features = feature_extractor_finetuned(images)

            finetuned_features.append(features.cpu().numpy())
            finetuned_labels.append(labels.numpy())

    # Восстанавливаем обученный head
    model.head = saved_head

    finetuned_features = np.vstack(finetuned_features) if finetuned_features else np.array([])
    finetuned_labels = np.concatenate(finetuned_labels) if finetuned_labels else np.array([])

    # Финальная оценка
    print("\n5. Финальное тестирование...")
    model.eval()
    all_predictions = []
    all_true_labels = []

    with torch.no_grad():
        for images, labels, _ in test_loader:
            images = images.to(device)
            labels = labels.to(device)
            outputs = model(images)
            _, predicted = torch.max(outputs, 1)
            all_predictions.extend(predicted.cpu().numpy())
            all_true_labels.extend(labels.cpu().numpy())

    final_accuracy = accuracy_score(all_true_labels, all_predictions)
    final_f1 = f1_score(all_true_labels, all_predictions, average='weighted')
    final_precision = precision_score(all_true_labels, all_predictions, average='weighted')
    final_recall = recall_score(all_true_labels, all_predictions, average='weighted')

    print(f"Финальная точность: {final_accuracy:.2%}")
    print(f"F1-score: {final_f1:.2%}")
    print(f"Precision: {final_precision:.2%}")
    print(f"Recall: {final_recall:.2%}")

    # Матрица ошибок
    cm = confusion_matrix(all_true_labels, all_predictions)
    print("\n Матрица ошибок:")
    print(cm)

    return {
        'model': model,
        'final_accuracy': final_accuracy,
        'final_f1': final_f1,
        'final_precision': final_precision,
        'final_recall': final_recall,
        'train_losses': train_losses,
        'val_accuracies': val_accuracies,
        'val_f1_scores': val_f1_scores,
        'original_features': original_features,
        'original_labels': original_labels,
        'finetuned_features': finetuned_features,
        'finetuned_labels': finetuned_labels,
        'confusion_matrix': cm
    }
finetune_results = finetune_swint_model(dataset, train_ratio=0.8, epochs=15,
                                       learning_rate=0.0001, device=device)
```

```python
def plot_feature_space(original_features, original_labels,
                       finetuned_features, finetuned_labels,
                       dataset_classes, title_prefix=""):
    """Визуализация пространства признаков до и после донастройки"""
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))

    colors = ['#FF6B6B', '#4ECDC4']

    # PCA до донастройки
    if len(original_features) > 0:
        pca_original = PCA(n_components=2, random_state=SEED)
        original_2d = pca_original.fit_transform(original_features)

        ax1 = axes[0]
        for class_idx, class_name in enumerate(dataset_classes):
            mask = original_labels == class_idx
            if np.any(mask):
                ax1.scatter(original_2d[mask, 0], original_2d[mask, 1],
                          color=colors[class_idx],
                          label=class_name,
                          alpha=0.7, s=50,
                          edgecolors='white', linewidth=0.5)

        explained_original = pca_original.explained_variance_ratio_.sum()
        ax1.set_title(f"{title_prefix}До донастройки\nОбъясненная дисперсия: {explained_original:.3f}",
                     fontsize=12, fontweight='bold')
        ax1.set_xlabel('PC1', fontsize=10)
        ax1.set_ylabel('PC2', fontsize=10)
        ax1.legend()
        ax1.grid(True, alpha=0.3)

    # PCA после донастройки
    if len(finetuned_features) > 0:
        pca_finetuned = PCA(n_components=2, random_state=SEED)
        finetuned_2d = pca_finetuned.fit_transform(finetuned_features)

        ax2 = axes[1]
        for class_idx, class_name in enumerate(dataset_classes):
            mask = finetuned_labels == class_idx
            if np.any(mask):
                ax2.scatter(finetuned_2d[mask, 0], finetuned_2d[mask, 1],
                          color=colors[class_idx],
                          label=class_name,
                          alpha=0.7, s=50,
                          edgecolors='white', linewidth=0.5)

        explained_finetuned = pca_finetuned.explained_variance_ratio_.sum()
        ax2.set_title(f"{title_prefix}После донастройки\nОбъясненная дисперсия: {explained_finetuned:.3f}",
                     fontsize=12, fontweight='bold')
        ax2.set_xlabel('PC1', fontsize=10)
        ax2.set_ylabel('PC2', fontsize=10)
        ax2.legend()
        ax2.grid(True, alpha=0.3)

    # t-SNE после донастройки (более качественная визуализация)
    if len(finetuned_features) > 0:
        try:
            tsne = TSNE(n_components=2, random_state=SEED, perplexity=min(30, len(finetuned_features)-1))
            finetuned_tsne = tsne.fit_transform(finetuned_features)

            ax3 = axes[2]
            for class_idx, class_name in enumerate(dataset_classes):
                mask = finetuned_labels == class_idx
                if np.any(mask):
                    ax3.scatter(finetuned_tsne[mask, 0], finetuned_tsne[mask, 1],
                              color=colors[class_idx],
                              label=class_name,
                              alpha=0.7, s=50,
                              edgecolors='white', linewidth=0.5)

            ax3.set_title(f"{title_prefix}t-SNE после донастройки",
                         fontsize=12, fontweight='bold')
            ax3.set_xlabel('t-SNE1', fontsize=10)
            ax3.set_ylabel('t-SNE2', fontsize=10)
            ax3.legend()
            ax3.grid(True, alpha=0.3)
        except:
            ax3.text(0.5, 0.5, "t-SNE не применим\n(недостаточно данных)",
                    ha='center', va='center', transform=ax3.transAxes)
            ax3.axis('off')

    plt.tight_layout()
    return fig
```

```python
def plot_training_history(train_losses, val_accuracies, val_f1_scores):
    """Визуализация процесса обучения"""
    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # График потерь
    axes[0].plot(train_losses, 'b-', linewidth=2)
    axes[0].set_xlabel('Эпоха', fontsize=10)
    axes[0].set_ylabel('Потери', color='b', fontsize=10)
    axes[0].tick_params(axis='y', labelcolor='b')
    axes[0].grid(True, alpha=0.3)
    axes[0].set_title('Потери при обучении', fontsize=12, fontweight='bold')

    # График точности и F1-score
    epochs = range(1, len(val_accuracies) + 1)
    axes[1].plot(epochs, val_accuracies, 'g-', linewidth=2, label='Точность (%)')
    axes[1].plot(epochs, val_f1_scores, 'r-', linewidth=2, label='F1-score (%)')
    axes[1].set_xlabel('Эпоха', fontsize=10)
    axes[1].set_ylabel('Метрика (%)', fontsize=10)
    axes[1].legend()
    axes[1].grid(True, alpha=0.3)
    axes[1].set_title('Метрики на валидации', fontsize=12, fontweight='bold')
    axes[1].set_ylim([0, 105])

    plt.tight_layout()
    return fig
```
